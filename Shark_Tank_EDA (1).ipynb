```python
# Importing necessary libraries
import pandas as pd
import numpy as np

# Load raw dataset
df = pd.read_csv("Shark Tank India.csv")
df.head()
```

```python
# Check data structure
df.info()

# Quick statistical overview
df.describe(include='all')
```

```python
# Total null values per column
df.isnull().sum()
```

```python
# Count and display duplicate rows
df.duplicated().sum()
df[df.duplicated()]
```

```python
# Check unique values per column to understand data types and quality
for col in df.columns:
    print(f"\nColumn: {col}")
    print(df[col].value_counts(dropna=False).head(5))
```

```python
# Make a copy for cleaning
df_cleaned = df.copy()
```

```python
# Remove leading/trailing whitespaces and convert to snake_case
df_cleaned.columns = df_cleaned.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
# Convert Episode number to integer if needed
df_cleaned['episode_number'] = df_cleaned['episode_number'].astype('Int64')
```

```python
# Check final cleaned structure
df_cleaned.info()

# Save to CSV
df_cleaned.to_csv("Cleaned_Shark_Tank_India.csv", index=False)
```
