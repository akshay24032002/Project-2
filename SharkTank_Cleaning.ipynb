# 🧼 Shark Tank India Data Cleaning Notebook

```
import pandas as pd
```

```
df = pd.read_csv("Shark Tank India.csv")
```

```
# Drop irrelevant or sparse columns
df.drop(columns=[
    'Anchor', 'Episode Title', 'Company Website', 'Cash Burn', 'Has Patents', 'Bootstrapped',
    'Part of Match off', 'Deal Has Conditions', 'Royalty Percentage', 'Royalty Recouped Amount',
    'Advisory Shares Equity', 'All Guest Names', 'Invested Guest Name'
], inplace=True)
```

```
# Rename key columns for simplicity
df.rename(columns={
    'Original Ask Amount': 'Ask Amount',
    'Total Deal Amount': 'Deal Amount',
    'Deal Valuation': 'Valuation',
    'Total Deal Equity': 'Equity',
    'Season Number': 'Season',
    'Startup Name': 'Startup'
}, inplace=True)
```

```
# Add Gender column from Male/Female counts
def determine_gender(row):
    if row['Male Presenters'] > 0 and row['Female Presenters'] == 0:
        return 'Male'
    elif row['Female Presenters'] > 0 and row['Male Presenters'] == 0:
        return 'Female'
    elif row['Female Presenters'] > 0 and row['Male Presenters'] > 0:
        return 'Mixed'
    else:
        return 'Unknown'

df['Gender'] = df.apply(determine_gender, axis=1)
```

```
# Calculate Profit % and Valuation in Crores
df['Profit %'] = (df['Deal Amount'] - df['Ask Amount']) / df['Ask Amount']
df['Valuation (Cr)'] = df['Valuation'] / 1e7
```

```
# Drop rows with missing Startup names and remove duplicates
df = df[df['Startup'].notna()].drop_duplicates(subset=['Startup'])
```

```
# Export cleaned dataset
df.to_csv("Cleaned_Shark_Tank_India_Updated.csv", index=False)
```
